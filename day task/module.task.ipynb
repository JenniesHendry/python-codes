```python
def largest_number(a,b,c):
    if a<b<c:
        print(c)
    if b<c<a:
        print(a)
    if c<a<b:
        print(b)
```

```python
largest_number(6,7,8)
```

```python
def calculator(a,b,o):
    if o=="*":
        print(a*b)
    if o=="+":
        print(a+b)
    if o=="-":
        print(a-b)
    if o=="/":
        print(a/b)
```

```python
calculator(4,5,"*")
calculator(10,45,"+")
calculator(12,16,"-")
calculator(17,40,"/")
```

```python
even=[]
odd=[]
for i in range(1,101):
    if i%2==0:
        even.append(i)
    else:
        odd.append(i)

print('even numbers are', even , "and there are", len(even),"even numbers")
print('odd numbers are' ,odd , "and there are", len(odd), "odd numbers")
```

```python
fizz=[]
buzz=[]
fizzbuzz=[]
for i in range(1,51):
    if i%3==0 and i%5==0:
        fizzbuzz.append(i)
    elif i%5==0:
        buzz.append(i)
    elif i%3==0:
        fizz.append(i)
print("fizz=multiples of 3",fizz)
print("buzz=multiples of 5",buzz)
print("fizzbuzz=multiples of 3,5",fizzbuzz)

```
