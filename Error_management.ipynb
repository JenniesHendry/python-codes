```python
try:
    a =2
    b=0
    print(a/b)
except:
    print("division error")
```

```python
try:
    n = 'o'
    res = 100 / n
    
except ZeroDivisionError:
    print("You can't divide by zero!")
    
except TypeError:
    print("only numbers are accepted")
except ValueError:
    print("Enter a valid number!")
    
else:
    print("Result is", res)
    
finally:
    print("Execution complete.")
```
